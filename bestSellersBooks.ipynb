Before analyzing data, we need to clean it

To remove duplicate rows:

```python
df_books.drop_duplicates()
```

Now I´m going to rename the Name column:

```python
df_books.rename(columns={'Name':'Title'})
```

```python
df_books['Price'].dtype
```

As you can see, the Price column have integer values, this may not be the best option, so I´m going to change it to float type:

```python
df_books['Price'] = df_books['Price'].astype(float)
```

Now we´re ready to begin the data analysis

Which autor have the most books on the bestsellers list?

```python
authors_counts = df_books['Author'].value_counts()
print(authors_counts)
```

Average rating by genre:

```python
rating = df_books.groupby('Genre')['User Rating'].mean()
print(rating)
```

Finally, we can export our analysis:

```python
# Top 10 authors
authors_counts.head(10).to_csv(r'C:\Users\Marcela Martínez\Desktop\top_authors.csv')

# Average rating by genre
rating.to_csv(r'C:\Users\Marcela Martínez\Desktop\rating.csv')
```
